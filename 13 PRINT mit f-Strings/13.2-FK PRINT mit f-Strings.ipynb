**Erkläre, was f-Strings in Python sind und welche Vorteile sie gegenüber der Ausgabe mit Kommas oder dem `+`-Operator bieten!**<br>
Was passiert, wenn man das `f` vor den Anführungszeichen vergisst? 

Vergleiche die Ausgabe von `print("Wert: {x}")` und `print(f"Wert: {x}")`.<br>

*f-Strings sind Strings mit einem f vor den Anführungszeichen. Variablen und Ausdrücke kommen direkt in {} rein, Python setzt beim Ausführen den Wert ein.*

*Vorteil: lesbarer, kein + und kein str() nötig, Formatierung gleich mit dabei.*

*Ohne f: normaler Text, die Klammern werden nicht ausgewertet. print("{x}") gibt tatsächlich {x} aus.*

**Beschreibe, was innerhalb der geschweiften Klammern `{}` eines f-Strings erlaubt ist!**<br>
Bestimme die genaue Bildschirmausgabe folgenden Codes und erkläre das Ergebnis:<br>

```python
    a = 7
    b = 4
    print(f"Die Rechnung {a} + {b} ergibt {a + b}!")
```

*In {} darf jeder Ausdruck stehen: Variablen, Rechnungen, Funktionsaufrufe, Methoden.*

*Ausgabe: Die Rechnung 7 + 4 ergibt 11! {a} und {b} werden durch ihre Werte ersetzt, {a + b} wird zuerst berechnet (summiert) und dann eingesetzt.*

**Erkläre die Formatierungsangabe bei Gleitkommazahlen:**<br>
Welche Bedeutung haben die einzelnen Bestandteile im Platzhalter `{wert:10.2f}`? Was bestimmt die `10`, was die `.2` und was das `f`?<br>

*10 ist die Mindestbreite, also 10 Zeichen inklusive Komma und Nachkommastellen: .2 bedeutet 2 Nachkommastellen (gerundet). f ist der Typ, also Gleitkommazahl.*

**Erkläre die Steuerzeichen für die Textausrichtung in f-Strings!**<br>
Nenne die Zeichen für linksbündig, zentriert und rechtsbündig und bestimme die Ausgabe von:<br>

```python
    text = "HTL"
    print(f"{text:*^9}")
```
<br>

*< ist linksbündig, ^ zentriert, > rechtsbündig. Ausgabe: `***HTL***`, weil 9 Zeichen Breite und HTL hat 3, bleiben 6 Sterne (3 links 3 rechts).*

**Analysiere und korrigiere folgende fehlerhafte f-Strings:**<br>
Erkläre jeweils, welcher Fehler vorliegt, und schreibe die korrigierte Zeile auf:<br>
a) `print(f"Trennlinie: {"-" * 20}")`<br>
b) `preis: float = 19.99; print(f"Preis: {preis:5d}")`<br>

*Innen und außen die gleichen Anführungszeichen: Python bricht den f-String beim inneren " ab, das gibt einen SyntaxError.*

*Richtig wäre print(f"Trennlinie: {'-' * 20}"), also innen einfache, außen doppelte Anführungszeichen.*

*d gibt's nur für Ganzzahlen, bei einem float kommt ValueError: Unknown format code 'd' for object of type 'float'.* 

*Richtig wäre in dem Fall print(f"Preis: {preis:5.2f}")*

**Erkläre das Konzept der String-Multiplikation (`*`) in Python!**<br>
Wie lässt sich die String-Multiplikation nutzen, um z. B. eine Trennlinie aus 40 Gleichheitszeichen (`=`) im Terminal zu erzeugen?<br>

*Ein String mal eine Ganzzahl wiederholt den String so oft.*

*Trennlinie: print("=" * 40) oder im f-String print(f"{'=' * 40}").*
