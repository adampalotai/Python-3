# 📘 PRINT mit f-Strings
---

Stell dir vor, du programmierst ein Spiel und möchtest am Ende eine persönliche Urkunde oder eine schöne Highscore-Tabelle anzeigen.  
Der Name der Spieler und ihre Punkte ändern sich ständig. Mit **f-Strings** kannst du Variablen so einfach in deinen Text einbauen wie bei einem Lückentext!

**`f-Strings`** (kurz für *formatted string literals*) sind die modernste und praktischste Möglichkeit in Python, **Variablen direkt in einen Text einzubauen**. 


### 🎯 Vorteile von f-Strings<br>
- Sie machen den Code **lesbarer** und **übersichtlicher**
- Du brauchst keine +-Zeichen zum Verbinden von Texten.
- Du kannst verschiedene Datentypen (z. B. int, float, str) einfach einfügen.
- Du kannst sogar Berechnungen oder Funktionen direkt in den Klammern verwenden!

Ein f-String beginnt mit dem Buchstaben **`f`** vor dem Anführungszeichen. Innerhalb des Textes kannst du dann **`Variablen in geschweiften Klammern ({})`** einfügen. Python ersetzt sie beim Ausführen durch den aktuellen Wert der Variablen.


`````python
# Variablen deklarieren
name:str  = "Anna"
alter:int = 17

# Klassisch: Mit Komma getrennt:
name = "Lukas"
alter = 18
print("Name:", name, "Alter:", alter)

# Mit f-String:
print(f"Name: {name} Alter: {alter}")
`````

Du kannst sogar *Rechnungen direkt in den Klammern* ausführen.

`````python
a: int = 5
b: int = 3
print(f"Die Summe von {a} und {b} ist {a + b}.")
`````

# 🟨 Ausgabe formatieren

Du kannst bei jeder Variablen angeben, 
- wie groß der Platz für die Ausgabe sein soll (Breite)
- wie viele Kommastellen ausgegeben werden sollen
- ob die Ausgabe linksbündig, rechtsbündig oder zentriert sein soll
- und welches Zeichen zum Füllen des leeren Platzes verwendet werden soll.

#### 🔹 Ganzzahlen rechtsbündig mit fester Breite
Das Kürzel **`d`** steht für *decimal integer* (Ganzzahl).

`````python
zahl:int = 42

print(f"Zahl: {zahl:5d}")  # Ausgabe mit 5 Stellen, rechtsbündig
`````

#### 🔹 Gleitkommazahlen mit Ausrichtung und Nachkommastellen
Das Kürzel **`f`** steht für *float* (Gleitkommazahl / Dezimalzahl).

`````python
wert:float = 12.34567

print(f"Wert: {wert:10.2f}")  # 10 Stellen Breite, 2 Nachkommastellen
`````

#### 🔹 Linksbündig, rechtsbündig und zentrierte Ausgaben

`````python
text:str = "Hi"

print(f"Links : {text:<10}!")
print(f"Zentriert: {text:^10}!")
print(f"Rechts: {text:>10}!")
`````

#### 🔹 Ausrichtung mit eigenen Füllzeichen

Standardmäßig füllt Python leeren Platz mit normalen Leerzeichen auf. Du kannst aber auch ein beliebiges anderes Zeichen (z. B. einen Punkt `.` oder einen Stern `*`) verwenden. 
Schreibe dieses Zeichen einfach direkt **vor** das Ausrichtungszeichen (`<`, `>`, `^`):

`````python
text:str = "Hi"

print(f"Mit Punkten: {text:.^10}!")  # Zentriert, mit Punkten gefüllt
print(f"Mit Sternen: {text:*>10}!")  # Rechtsbündig, mit Sternen gefüllt
`````

#### 🔹 Zahlen mit führenden Nullen

`````python
nummer:int = 42

print(f"Artikel-Nr: {nummer:05d}")  # gibt 00042 aus
`````

| Zweck | Syntax im f-String | Beispiel | Ergebnis (Beispiel) |
|---|---|---|---|
| Variable einfügen | `{variable}` | `f"Name: {name}"` | `Name: Lukas` |
| Ausdruck/Berechnung einfügen | `{ausdruck}` | `f"Summe: {a + b}"` | `Summe: 8` |
| Ganzzahl, rechtsbündig mit Breite | `{zahl:5d}` | `f"{42:5d}"` | `'   42'` |
| Ganzzahl mit führenden Nullen | `{zahl:05d}` | `f"{42:05d}"` | `00042` |
| Float mit Breite und Nachkommastellen | `{wert:10.2f}` | `f"{12.34567:10.2f}"` | `'     12.35'` |
| Text linksbündig | `{text:<10}` | `f"{'Hi':<10}"` | `'Hi        '` |
| Text zentriert | `{text:^10}` | `f"{'Hi':^10}"` | `'    Hi    '` |
| Text rechtsbündig | `{text:>10}` | `f"{'Hi':>10}"` | `'        Hi'` |
| Eigenes Füllzeichen + zentriert | `{text:.^10}` | `f"{'Hi':.^10}"` | `....Hi....` |
| Eigenes Füllzeichen + rechtsbündig | `{text:*>10}` | `f"{'Hi':*>10}"` | `********Hi` |
| Zeichen wiederholen (String-Multiplikation) | `"zeichen" * anzahl` | `"-" * 30` | `------------------------------` |

##### ✨ Merksatz zur Format-Syntax
`{wert:[füllzeichen][ausrichtung][breite][.präzision][typ]}`

- **Ausrichtung:** `<` (links), `^` (zentriert), `>` (rechts)  

- **Typ:** `d` (Ganzzahl), `f` (Gleitkommazahl)

# 🔁 Zeichen wiederholen (String-Multiplikation)

Manchmal möchtest du Trennlinien oder Rahmen zeichnen, zum Beispiel `--------------------`.
Statt das Zeichen 40-mal selbst in den Code zu tippen, kannst du einen Text (String) in Python einfach mit einer Zahl multiplizieren!

**Syntax:** `"Zeichen" * Anzahl`

`````python
# Wiederhole einen Strich 30-mal
linie = "-" * 30
print(linie)

# Das geht auch direkt im f-String eingebaut:
print(f"Deko: {'★' * 5} Willkommen {'★' * 5}")
`````

# 🟥 Typische Anfängerfehler & wie du sie vermeidest


1. **Das `f` vor den Anführungszeichen vergessen:**  
   Fehlt das `f` vor dem String, behandelt Python den Text als gewöhnlichen Text. Die geschweiften Klammern werden nicht ausgewertet!  
   *Falsch:* `print("Hallo {name}")` $\rightarrow$ Ausgabe: `Hallo {name}`  
   *Richtig:* `print(f"Hallo {name}")` $\rightarrow$ Ausgabe: `Hallo Anna`

1. **Falsche Klammern verwendet:**  
   Für die Platzhalter in f-Strings dürfen ausschließlich **geschweifte Klammern `{}`** verwendet werden – keine runden `()` oder eckigen `[]` Klammern.  
   *Falsch:* `print(f"Alter: (alter)")` $\rightarrow$ Ausgabe: `Alter: (alter)`  
   *Richtig:* `print(f"Alter: {alter}")` $\rightarrow$ Ausgabe: `Alter: 17`

1. **Gleiche Anführungszeichen innen und außen verschachtelt:**  
   Wird innerhalb der geschweiften Klammern `{}` ein String mit denselben Anführungszeichen wie außen verwendet, bricht Python den f-String vorzeitig ab.  
   *Falsch:* `print(f"Trennlinie: {"-" * 20}")` $\rightarrow$ `SyntaxError: invalid syntax`  
   *Richtig:* Unterschiedliche Anführungszeichen kombinieren: `print(f"Trennlinie: {'-' * 20}")`

1. **Typkonflikt beim Formatbezeichner (`d` statt `f`):**  
   Das Kürzel `d` steht ausschließlich für Ganzzahlen (`int`). Wird `d` bei einer Kommazahl (`float`) angegeben, stürzt das Programm ab!  
   *Falsch:* `preis = 9.99; print(f"{preis:5d}")` $\rightarrow$ `ValueError: Unknown format code 'd' for object of type 'float'`  
   *Richtig:* `print(f"{preis:5.2f}")`

1. **Reihenfolge der Formatierungsangaben vertauscht:**  
   Die Reihenfolge nach dem Doppelpunkt ist streng vorgegeben: `[Füllzeichen][Ausrichtung][Breite][.Nachkommastellen][Typ]`.  
   *Falsch:* `print(f"{text:10<}")` oder `print(f"{zahl:d5}")`  
   *Richtig:* `print(f"{text:<10}")` und `print(f"{zahl:5d}")`

# 🟩 Zusammenfassung: Das Wichtigste auf einen Blick

- Ein **f-String** beginnt immer mit einem vorangestellten `f` unmittelbar vor den Anführungszeichen: `f"..."` oder `f'...'`.  

- **Variablen und Ausdrücke** werden in geschweifte Klammern `{}` eingefügt und beim Ausführen automatisch als Text dargestellt.  
- **Berechnungen** können direkt innerhalb der geschweiften Klammern durchgeführt werden (z. B. `{a + b}` oder `{breite * hoehe}`).  
- **Format-Spezifikation** erfolgt nach einem Doppelpunkt: `{wert:[füllzeichen][ausrichtung][breite][.präzision][typ]}`.  
- **String-Multiplikation:** Mit `"Zeichen" * anzahl` können Zeichenketten schnell wiederholt werden (z. B. `"-" * 40` für Trennlinien).
