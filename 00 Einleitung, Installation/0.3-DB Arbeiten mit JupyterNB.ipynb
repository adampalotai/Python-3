# <img src="img/icon_vscode.png" alt="VS Code" width="48" style="vertical-align: middle; margin: 0 6px 2px 4px;"> Arbeiten mit Jupyter Notbooks

**``Jupyter-Notebboks``** sind interaktive Arbeitsblätter welche wir im SEW-Unterricht verwenden.

Sie enthalten 
- Texte und Erklärungen, 
- Programmbeispiele, 
- Aufgaben und Übungen, welche selbst zu lösen sind, 
- Lösungen und 
- auch Fragen, welche zu beantworten sind

Mit diesem **``Jupyter-Notebboks``** üben wir die ersten Schritte.

Mit der Taste **ENTER** kannst du einen **``Block``** zum bearbeiten öffnen und selbst etwas hineinschreiben.<br>
Wenn du damit fertig bist kannst du mit **SHIFT-ENTER** das Bearbeiten beenden und den **``Block``** wieder normal anzeigen.
