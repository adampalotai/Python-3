# 🟨 Strings in <code>print()</code> aufteilen
Oft reicht es nicht aus einfach nur einen String mit <code>print()</code> auszugeben. 

`````python
# Beispiel: Berechnung der Gesamtlänge von zwei Längenangaben in Zentimetern
print("Die Gesamtlänge von 567cm + 1789cm ist: 2356cm")
`````

Wollen wir das Ergebnis vom Computer berechnen lassen, müssen wir 
- zuerst den Text ausgeben
- dann rechnen lassen und 
- danach noch die Einheit ausgeben.

Verbunden werden die einzelnen Teile mit einem Komma.

`````python
# Beispiel: Berechnung der Gesamtlänge von zwei Längenangaben in Zentimetern
# und aufteilen des Strings in mehrere Teile, um die Berechnung den Computer zu überlassen
print("Die Gesamtlänge von 567cm + 1789cm ist:", 567 + 1789 ,"cm")
`````

## 🔸 Der Parameter `end` – Was kommt am Ende der Ausgabe?
Standardmäßig fügt `print()` **einen Zeilenumbruch** am Ende hinzu (`\n`). Du kannst das ändern.

`````python
print("Erste Zeile", end=" - ")
print("Zweite Zeile")  # Beide in einer Zeile
`````

## # 👉 ---> Dein Code Der Parameter `sep` – Trennzeichen zwischen Argumenten
Standardmäßig trennt `print()` mehrere Argumente mit einem **Leerzeichen**. Du kannst das ändern.

`````python
print("Tag", "Nacht", "Sonne")         # Standard (mit Leerzeichen)
print("Tag", "Nacht", "Sonne", sep=" | ")   # Benutzerdefiniertes Trennzeichen
`````

# 🟨 Steuerzeichen (Escape-Sequenzen)

In einer <code>print()</code> Anweisung hat ein Backslash `\` (verkehrter Schrägstrich) eine besondere Bedeutung.

Gemeinsam mit dem nächsten Zeichen bildet er einen Steuercode für die Ausgabe - eine sogenannte `ESC-Sequenz`.  
Man kann damit zum Beispiel
- eine neue Zeile oder
- eine Tabulator

in den Text einfügen.



<div style="background-color:rgba(0, 255, 0, 0.1); border-left: 4px solid #4CAF50; padding: 30px;">

##### ✨ Die wichtigsten Steuerzeichen sind
- **`\n`** → Zeilenumbruch (Newline)

- **`\t`** → Tabulator (Einrückung)
- **`\\`** → Backslash selbst ausgeben
- **`\"`** → doppeltes Anführungszeichen

</div>

`````python
# Zeilenumbruch einfügen
print("Zeile 1\nZeile 2")   # \n fügt einen Zeilenumbruch ein, sodass der Text in zwei Zeilen ausgegeben wird

# Tabulator einfügen 
print("Name:\t\tAnna")      # \t fügt einen Tabulator ein, sodass der Text eingerückt wird
`````

<div style="background-color:rgba(255, 165, 0, 0.1); border-left: 4px solid #ffa500; padding: 30px; margin: 10px 0;">

##### ✨ Merke:
Ein Backslash `\` hat in <code>print()</code> eine besondere Bedeutung, er ist der Beginn einer `ESC-Sequenz`.

*Willst du `einen Backslash` ausgeben, musst du für jeden Backslash in einer print-Anweisung `zwei Backslash` schreiben!*

```python
    print("So gibt man einen Backslash \\ auf dem Monitor aus!")
```

ergibt als Ausgabe:

```text
    So gibt man einen Backslash \ auf dem Monitor aus!
```

</div>

`````python
# Backslash korrekt ausgeben
print("Pfad: C:\\Users\\Anna")     # Pfad mit doppeltem Backslash ausgeben, damit der Backslash korrekt dargestellt wird
`````

# 🟨 Strings auf mehrere Zeilen aufteilen und verbinden

Oft ist es sinnvoll bei der Ausgabe Strings in mehrere Zeilen aufzuteilen und mit `+` zu verbinden.  
Dadurch wird die Übersicht und Lesbarkeit eines Programmes stark verbessert

`````python
# Beispiel: Eine print-Anweisung, die mehrere Zeilen mit + verbindet und \n verwendet
print("Es war einmal ein kleiner Wurm,\n"     +
      "der wollte schnell zu einem Turm.\n"   +
      "Doch auf dem Weg schlug er ein Rad,\n" +
      "und lachte laut: 'Mir ist nicht Fad!'")
`````
