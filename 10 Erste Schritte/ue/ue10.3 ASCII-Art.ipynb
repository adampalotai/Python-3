# 🦄  `Übungen`:  ASCII-Art und ESC-Sequenzen
---

**ASCII-Art** ist eine Kunstform, bei der Bilder, Illustrationen oder Schriftzüge ausschließlich mithilfe von Textzeichen (Buchstaben, Ziffern und Sonderzeichen) dargestellt werden. 

Früher, als Computerbildschirme und Drucker noch keine hochauflösenden Grafiken anzeigen konnten, war dies die einzige Möglichkeit, visuelle Darstellungen am Computer zu erzeugen.

Hier findest du viele weitere Beispiele und Vorlagen:
