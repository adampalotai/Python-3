# 🟨 Elemente eines Python-Programms

Ein Python-Programm besteht aus vielen **`Anweisungen`**. Sie werden auch als **`Statement`** bezeichnet.  
- Sie sind *Befehle* (Codezeilen), die dem Computer sagen, was er zu tun hat und
- werden normalerweise *von oben nach unten der Reihe nach* ausgeführt. 

Anweisungen können 
- Werte berechnen, 
- in Variablen speichern, 
- Funktionen aufrufen oder 
- Ausgaben auf dem Bildschirm erzeugen.

Ein Programm besteht dabei aus vielen Tausenden Codezeilen.

````python
""" ----------------------------------------------------------------------------------
 Programmname: PixelQuest - Highscore-Rechner
 Erstellt von: HTL Traun
 Beschreibung: Begrüßt Spieler, sammelt Münzen ein und berechnet den Punktestand.
------------------------------------------------------------------------------------- """

# ------------------------------------------------------------------------------------
# 1. STARTWERTE (Der Computer merkt sich grundlegende Spieldaten)
spiel_titel:str  = "PixelQuest 2026"
start_punkte:int = 100

# ------------------------------------------------------------------------------------
# 2. EINGABE (Der Computer fragt den Spieler)
spieler_name = input("Wie lautet dein Gamer-Tag? ")
gesammelte_muenzen = int(input("Wie viele Goldmünzen hast du gesammelt? "))

# ------------------------------------------------------------------------------------
# 3. VERARBEITUNG (Der Computer rechnet: Jede Münze ist 10 Punkte wert)
muenzen_punkte = gesammelte_muenzen * 10
gesamt_punkte = start_punkte + muenzen_punkte

# ------------------------------------------------------------------------------------
# 4. AUSGABE (Das Ergebnis wird auf dem Bildschirm präsentiert)
print()
print("========================================")
print("      WILLKOMMEN BEI", spiel_titel)
print("========================================")
print("Spieler:", spieler_name)
print("Start-Bonus:", start_punkte, "Punkte")
print("Münz-Bonus:", muenzen_punkte, "Punkte")
print("----------------------------------------")
print("DEIN GESAMTER SCORE:", gesamt_punkte)
print("========================================")
print("Level 1 erfolgreich abgeschlossen! 🚀")
````

# 🟨 Kommentare in Python

Damit Menschen Programme besser verstehen können, gibt es in jedem **`Programm`** auch **`Kommentare`**.<br>  
**`Kommentare`** sind beliebige Texte, welche für den Programmierer gedacht sind und vom Computer ignoriert werden.

Kommentare beginnen in Python mit einem <kbd>**#**</kbd>:

```python
    # Dieser Kommentar kann mehrere Zeilen lang sein.
    # Jede Kommentarzeile beginnt erneut mit einem #.
```

Dreifache Anführungszeichen (`"""..."""`) erzeugen einen mehrzeiligen Text und werden genutzt, wenn längere Beschreibungen eingefügt werden sollen.

```python
    """ ----------------------------------------------------------------------------------
    Programmname: Grundelemente in Python
    Beschreibung: Zeigt Variablen, Datentypen,
                Eingabe, Verarbeitung und Ausgabe.
    -------------------------------------------------------------------------------------
    """
```


````python
# Dieser Kommentar beschreibt was die folgenden Zeilen machen.
breite = 23 + 10      # Addiere die Werte und speichere das Ergebnis in breite.
laenge = breite * 2   # Die Laenge ist doppelt so gross wie die Breite.
print(laenge)
````

# 🟨 Texte in Python

Alles zwischen passenden Anführungszeichen wird als **`Text`** behandelt. Texte heißen auch **`Strings`**. Du kannst einfache (`'...'`) oder doppelte (`"..."`) Anführungszeichen verwenden. Anfang und Ende müssen zusammenpassen.

```python
    "Hello World!"  # Das ist für Python ein Text; dieser wird STRING für Zeichenkette genannt
    'Hallo'         # Ebenfalls ein String
    "laenge = 4"    # Schaut aus wie eine Anweisung! Ist aber ein Text und wird deshalb nicht als Anweisung ausgeführt
    laenge = 4      # Anweisung; speichert die Zahl 4 in der Variablen laenge
```



````python
# Hier wird die Länge ausgegeben, aber es ist nur ein Text. Die Variable laenge wird nicht ausgegeben.
print("laenge = 4")  
````

````python
# Texte können mit dem Pluszeichen (+) aneinandergefügt werden.
# Das werden wir später häufiger verwenden, um z.B. Variablen in Texten auszugeben.
print("Hallo," + " zum Unterricht an der" + " HTL-Traun" + " im Fach" + " SWT" + ".")
````

# 🟨 Etwas auf dem Bildschirm ausgeben

Eine Ausgabe zeigt Informationen für Benutzerinnen und Benutzer an. In Python geschieht das mit:

```python
    print(...)
```

<code>print()</code> wechselt nach der Ausgabe automatisch in eine neue Zeile. Mit dem benannten Argument `end` kann dieses Verhalten verändert werden:

```python
    print("Prof.", end=" ")
    print("Gaisberger")
```

### Beispiele für Ausgaben

````python
# Ausgabe der Zahl 5
print(5)
````

````python
# Ausgabe der Summe zweier Zahlen
print(123 + 34)
````

````python
# Berechnungen mit Kommazahlen und das Ergebnis ausgeben
print(3.5 + 78.32 - 14.893)
````

````python
# Die vier Grundrechenarten und Modulo
print(20 + 5)
print(20 - 5)
print(20 * 5)
print(20 / 5)   # / liefert in Python immer eine Kommazahl (float).
print(20 % 5)   # % berechnet den Rest einer Division.

print(45.3 / 22)
````

````python
# Häufig wird beim Programmieren der Rest einer Division benötigt:
# Dies nennt man Modulo-Operation und wird mit dem Zeichen % angegeben
print(24 % 7)
````

````python
# Ausgabe eines Textes
# Achtung der Text muss dabei zwischen 2 Hochkomma geschrieben werden
print("Spaß mit SWT")
````

````python
# Ausgabe mit und ohne Zeilenumbruch
print("Prof.", end=" ")         # print(...) bleibt in der gleichen Zeile und fügt ein Leerzeichen ein.
print("Gaisberger", end=" ")    # print(...) bleibt in der gleichen Zeile und fügt ein Leerzeichen ein.
print("Alois")                  # print(...) wechselt in die nächste Zeile.
print("HTL Traun")              # print(...) wechselt in die nächste Zeile.
````

````python
# Mehrere Zeilen ausgeben
# Jeder print-Befehl erzeugt eine neue Zeile.
print("HTL Traun")
print("Prof. Gaisberger")
print("Software-Entwicklung")
````
