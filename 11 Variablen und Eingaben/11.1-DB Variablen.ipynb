In Mathematik hast du sicherlich schon häufig solche und andere Gleichungen gelöst:


- $x =  {\displaystyle 20 \over \displaystyle 5}  + 17  $

# 🟨 Deklarieren einer Variablen

Um eine Variable in einem Programm verwenden zu können, muss sie zuerst `DEKLARIERT` werden. Das heißt, sie muss dem System bekannt gegeben werden, damit Python sie verwenden kann.

> Ähnlich wie: *`Du brauchst zuerst eine Schachtel bevor du etwas hineingeben kannst!`*

````python
myName   = "Bob"             # Variable myName deklariert und mit dem Wert "Bob" initialisiert
myAge    = 17                # Variable myAge deklariert und mit dem Wert 17 initialisiert
myWeight = 70.5              # Variable myWeight deklariert und mit dem Wert 70.5 initialisiert
isHappy  = True              # Variable isHappy deklariert und mit dem Wert True initialisiert
````

Nun kannst du mit diesen Variablen weiterarbeiten. Du kannst sie zum Beispiel in der Konsole ausgeben.

````python
print("Der Wert in der Variable myName ist ", myName, "!")  # Ausgabe des Werts von myName
````

## Datentypen von Variablen

Wie du bei den Beispielen sehen kannst, haben die Werte der Variablen unterschiedliche Eigenschaften:
- *myName*: Ist ein Text (Datentyp `str` / String)
- *myAge*: Ist eine Ganzzahl ohne Komma (Datentyp `int` / Integer)
- *myWeight*: Ist eine Zahl mit Komma (Datentyp `float` / Gleitkommazahl)
- *isHappy*: Ist ein Wahrheitswert, also entweder Ja (`True`) oder Nein (`False`) (Datentyp `bool` / Boolean)

> Man nennt diese unterschiedlichen Eigenschaften auch **`Datentypen`**.

Wir unterscheiden folgende grundlegende Datentypen:

<div style="background-color:rgba(0, 255, 0, 0.1); border-left: 4px solid #4CAF50; padding: 30px;">

Name   | Datentyp
-------| ---------
`str`  | *STRING*: Steht für einen Text
`int`  | *INTEGER*: Steht für eine Zahl OHNE Komma (Natürliche Zahl)
`float`| *FLOAT*: Steht für eine Zahl MIT Komma (Reele Zahl)
`bool` | *BOOLEAN*: Steht für einen Wahrheitswert JA/NEIN

</div>

<div style="background-color:rgba(255, 165, 0, 0.1); border-left: 4px solid #ffa500; padding: 30px; margin: 10px 0;">

**In PYTHON ist es `NICHT` unbedingt nötig, Datentypen anzugeben.**

Es gibt jedoch viele Gründe, warum `Datentypen` beim Programmieren einen großen Einfluss besitzen. 
- Zum Beispiel schützen sie dich vor Programmierfehlern.
- Machen Programme effizienter
- und vieles mehr

<mark>Deshalb werden wir vorerst beim Deklarieren JEDER Variablen auch ihren DATENTYP angeben!</mark>

Deshalb lautet die korrekt Deklaration der obigen Variablen folgendermaßen:

````python
myName:str     = "Bob"             # Variable myName deklariert und mit dem Wert "Bob" initialisiert
myAge:int      = 17                # Variable myAge deklariert und mit dem Wert 17 initialisiert
myWeight:float = 70.5              # Variable myWeight deklariert und mit dem Wert 70.5 initialisiert
isHappy:bool   = True              # Variable isHappy deklariert und mit dem Wert True initialisiert
````

# 🟨 Schreibweise von Variablennamen
Für die Schreibweise von Variablennamen müssen bestimmte feste syntaktische Regeln von Python eingehalten werden, ansonsten stürzt das Programm mit einer Fehlermeldung ab.

Zusätzlich gibt es Richtlinien für schönen und lesbaren Code, die wir **``Namenskonventionen``** nennen. Es ist auch wichtig, dass der Name einer Variable aussagekräftig ist (also beschreibt, was darin gespeichert ist).

<div style="background-color:rgba(0, 255, 0, 0.1); border-left: 4px solid #4CAF50; padding: 30px;">

##### ✨ Namenskonventionen & Regeln:

- Der erste Buchstabe wird **immer klein** geschrieben (z. B. `mein_alter` oder `meinAlter`).

- Variablennamen müssen mit einem **Buchstaben** oder einem **Unterstrich `_`** beginnen. Sie dürfen niemals mit einer Zahl beginnen!
- Groß- und Kleinschreibung wird strikt unterschieden (`alter` und `Alter` sind zwei verschiedene Variablen).
- Es dürfen **keine Leerzeichen** und **keine Sonderzeichen** (wie `?`, `!`, `*`, `-`, etc.) verwendet werden.
- **Umlaute (ä, ö, ü, ß):** werden als `ae, oe, ue, und ss` geschrieben

---

*Obwohl Python 3 Umlaute im Variablennamen technisch erlaubt, verwenden wir in der professionellen Programmierung **ausschließlich englische Buchstaben (a-z) ohne Umlaute** (z. B. `groesse` statt `größe`), um Kompatibilitätsprobleme zu vermeiden.*

</div>

# 🟨 Deklarieren und Initialisieren




Für Programmier-Anfänger sind diese beiden Begriffe oft verwirrend. Am besten versteht man sie mit unserer **Schachtel-Metapher**:

#### 1. Deklarieren (Die Schachtel vorbereiten)
Wenn wir eine Variable **deklarieren**, reservieren wir einen Platz im Speicher des Computers. Wir sagen dem Computer, wie unsere Variable heißen soll und welche Art von Daten hineingehören (der Datentyp).

- **Metapher:** Du nimmst eine leere Schachtel und klebst ein Schild darauf, z. B. *"Hier nur Ganzzahlen (`int`)"* und gibst ihr den Namen `schuhgroesse`.

  ```python
    schuhgroesse: int
  ```
  *Wichtig:* Die Schachtel ist jetzt noch **völlig leer**. Wenn wir in Python versuchen würden, eine nur deklarierte Variable sofort auszugeben, gäbe es einen Fehler, weil noch gar nichts darin liegt!

#### 2. Initialisieren (Den ersten Inhalt hineinlegen)
Das **Initialisieren** ist die allererste Wertzuweisung. Hier bekommt die deklarierte Variable ihren allerersten Wert (Anfangswert).

- **Metapher:** Du legst zum ersten Mal ein Paar Schuhe der Größe 43 in deine beschriftete Schachtel hinein.

  ```python
    schuhgroesse = 43
  ```

#### 3. Der Unterschied auf einen Blick
<div style="background-color:rgba(255, 165, 0, 0.1); border-left: 4px solid #ffa500; padding: 30px; margin: 10px 0;">

| Begriff | Was passiert? | Metapher | Code-Beispiel |
| :--- | :--- | :--- | :--- |
| **Deklarieren** | Speicherplatz reservieren & Datentyp festlegen | Leere Schachtel beschriften | `alter: int` |
| **Initialisieren** | Den allerersten Wert zuweisen | Inhalt das erste Mal hineinlegen | `alter = 17` |
| **Zuweisen** | Einen neuen Wert hineinspeichern | Alten Inhalt durch neuen ersetzen | `alter = 18` |


</div>

<div style="background-color:rgba(0, 255, 0, 0.1); border-left: 4px solid #4CAF50; padding: 30px;">

##### ✨ Merke:

Du kannst in Python das Deklarieren und Initialisieren auch in **einem einzigen Schritt** durchführen. Das ist meistens kürzer und übersichtlicher:

```python
  alter: int = 17  # Deklariert (int) und initialisiert (17) in einer Zeile!
```

# 🟨 Zuweisen und Verwenden von Variablen
Nachdem eine **``Variable deklariert``** wurde, kann sie in weiteren **``Anweisungen``** verwendet werden.

> Wir können
> - einer Variable Werte ***zuweisen***
> - mit ihr ***rechnen*** und
> - sie ***ausgeben***

## Variablen Werte zuweisen
Nach dem reinen Deklarieren (Festlegen des Typs, z. B. `laenge: int`) besitzt eine Variable in Python noch keinen Standardwert (nicht einmal `0`!). 

Sie existiert im Speicher erst dann richtig, wenn wir ihr einen ersten Wert **``zuweisen``**.
Eine **``Zuweisung``** eines Wertes erfolgt immer mit dem **``Zeichen  =``**.

<div style="background-color:rgba(0, 255, 0, 0.1); border-left: 4px solid #4CAF50; padding: 30px;">

##### ✨ Merke:

Die Variable welche den Wert erhalten soll, steht IMMER <mark>**LINKS**</mark> von **`` = ``** und der Wert <mark>**RECHTS**</mark>

- **`Zielvariable` = Wert**


</div>

#### Diese Zuweisungen sind RICHTIG

````python
laenge: int                 # Deklarieren der Variable laenge

laenge = 5                  # RICHTIG: Variable LINKS, Wert RECHTS
laenge = 7 + 2              # 7 + 2 ergibt 9 und wird der Variable laenge zugewiesen

print(f"Die Länge ist {laenge}")
````

````python
breite: int                 # Deklarieren der Variable breite

breite = 2 * laenge         # RICHTIG: Variable LINKS, Wert RECHTS
                            # laenge ist 9 und mit 2 multipliziert ergibt dies 18.

print(f"Die Breite ist {breite}")
````

#### Diese Zuweisungen sind FALSCH

````python
5 = breite                 # FALSCH: Zielvariable muss LINKS stehen, Wert RECHTS.
laenge = = 10              # FALSCH: Zwei Zuweisungszeichen hintereinander.

5 = breite                 # FALSCH: Zielvariable muss LINKS stehen, Wert RECHTS.
5 + 2 = breite             # FALSCH: Zielvariable muss LINKS stehen, Wert RECHTS.
2 * laenge = breite        # FALSCH: Zielvariable muss LINKS stehen, Wert RECHTS.
````

## Variable initialisieren
Die erste Zuweisung eines Wertes zu einer Variable wird auch **``Initialisieren``** genannt, da sie dadurch ihren **Anfangswert** erhält.

````python
hoehe: int                  # Deklarieren einer Variable hoehe für Zahlen OHNE Komma
                            # Achtung: Sie hat noch keinen Wert! Wenn wir sie jetzt ausgeben würden, gäbe es einen Fehler.

hoehe = 10                  # Nun INITIALISIEREN wir die Variable hoehe mit einem Wert
                            # und können diesen Wert danach ausgeben
print(f"Die Höhe ist {hoehe}")
````

Das **``Deklarieren``** und **``Initialisieren``** einer Variable kann auch in einem Schritt erfolgen.


````python
hoehe:int = 10              # Deklarieren einer Variable hoehe für Zahlen OHNE Komma
                            # und gleichzeitig mit einem Anfangswert INITIALISIEREN
                            # und diesen Wert ausgeben
print(f"Die Höhe ist {hoehe}")
````

Nun können wir auch mit dieser Variablen rechnen und das Ergebnis wieder einer Variablen **``zuweisen``**.

````python
# Überlege zuvor was dieses Programm ausgeben wird
hoehe = hoehe + 19             # Addiere zu hoehe die Zahl 19

print(f"Jetzt ist die Höhe {hoehe}")
````
