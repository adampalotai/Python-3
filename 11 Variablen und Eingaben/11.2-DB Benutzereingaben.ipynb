# 📘 INPUT - Eingaben des Benutzers einlesen
---


Um eine Eingabe vom Benutzer anzufordern, gibt es in Python den Befehl <code>input()</code>

````python
    alter = input("Bitte gib dein Alter in Jahren ein: ")
````

- Zuerst wird der Text "Bitte gib dein Alter in Jahren ein: " auf dem Bildschirm ausgegeben 
- nun wartet das Programm so lange, bis der Benutzer etwas eingibt und 
- speichert danach diese Eingabe in der Variablen <code>alter</code>.


`Es ist wichtig, den Text bei input() anzugeben, sonst weiß der Benutzer nicht, was er tun soll.`

#### Was geschieht, wenn der Benutzer keine Zahl eingibt? 


`````python
# Benutzer zur Eingabe auffordern
alter = input("Bitte geben Sie Ihr Alter ein: ")

print(f"Ihr Alter ist: {alter}")
`````

<div style="background-color:rgba(255, 165, 0, 0.1); border-left: 4px solid #ffa500; padding: 30px; margin: 10px 0;">

##### ✨ Merke:

**Für Python ist das kein Problem, es arbeitet einfach weiter ...**

Für dein Programm ist es schon ein Problem, denn 
- es erwartet ja eine Zahl, *nämlich das Alter in Jahren* 
- und soll später damit gerechnet werden, **stürzt dein Programm ab.**


</div>

`````python
# Berechne das Alter in 10 Jahren
alter_in_10_jahren = alter + 10
`````

**`Das ist mit ein Grund warum wir Datentypen bei jeder Variablen angeben.`**

⚠️ Für Python ist jede Eingabe zuerst einmal ein String `str`!

Deshalb MUSS jede Eingabe in den Datentyp umgewandelt werden, der erwartet wird.

`````python
# Korrekte Verwendung von input() und Umwandlung in Integer
alter:int = int(input("Bitte geben Sie Ihr Alter ein: "))

alter_in_10_jahren:int = alter + 10

print(f"In 10 Jahren werden Sie {alter_in_10_jahren} Jahre alt sein.")
`````

<div style="background-color:rgba(255, 165, 0, 0.1); border-left: 4px solid #ffa500; padding: 30px; margin: 10px 0;">

##### ✨ Merke:

Gibt der Benutzer jetzt etwas anderes ein,

- stürzt das Programm trotzdem ab, aber
- `dort wo der Fehler aufgetreten ist und nicht irgendwann später`

# 🟨 Eingaben für verschiedene Datentypen



### Eingabe einer Zahl OHNE Komma - `int`

`````python
# Benutzer zur Eingabe auffordern und Eingabe in Integer umwandeln
anzahl_schueler: int = int(input("Wie viele Schüler sind heute anwesend? "))

print(f"Heute sind {anzahl_schueler} Schüler anwesend.")
`````

### Eingabe einer Zahl MIT Komma - `float`

`````python
# Benutzer zur Eingabe auffordern und Eingabe in Float umwandeln
groesse: float = float(input("Wie groß bist du in Metern: "))

print(f"Du bist {groesse} Meter groß.")
`````

### Eingabe eines Textes - `str`

`````python
# Beispiel für eine Texteingabe (str) mit konkretem Bezug zur Schule
# Strings müssen nicht extra umgewandelt werden, da input() bereits einen String zurückgibt
lieblingsfach: str = input("Welches Fach magst du in der Schule am meisten? ")

print(f"Dein Lieblingsfach ist {lieblingsfach}.")
`````
