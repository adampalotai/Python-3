### ➡️ *Aufgabe:* Das Rechteck

Lies die Länge und die Breite eines Rechtecks ein.  
Berechne den Flächeninhalt des Rechtecks und gib das Ergebnis mit <code>print()</code> aus.  

*Tipp: Deklariere für Länge, Breite und Fläche eine `Variable` mit einem geeigneten `Datentyp`.*


> <div style="background-color:rgba(0, 0, 250, 0.1)">  
>
> **Erwartete Ausgabe:**
> ````Text
>  Gib die Länge des Rechtecks ein: 5.4
>  Gib die Breite des Rechtecks ein: 3.2
>
>  Die Länge des Rechtecks ist: 5.4
>  Die Breite des Rechtecks ist: 3.2
>  Die Fläche des Rechtecks ist: 17.28
> ````

`````python
# 👉 ---> Dein Code
`````

<details><summary>💡 Lösung</summary>

```python
    # Einlesen und Initialisieren der Variablen
    laenge: float = float(input("Gib die Länge des Rechtecks ein: "))
    breite: float = float(input("Gib die Breite des Rechtecks ein: "))

    # Berechnen des Ergebnisses
    flaeche: float = laenge * breite

    # -------------------------------------------------------------
    # Ausgabe der Variablen
    print(f"Die Länge des Rechtecks ist: {laenge}")
    print(f"Die Breite des Rechtecks ist: {breite}")
    print(f"Die Fläche des Rechtecks ist: {flaeche}")
```

### ➡️ *Aufgabe:* Die Prüfung

Lies die Punkte eines Schülers aus drei Prüfungen ein.  

Berechne den Durchschnitt dieser drei Noten und gib das Ergebnis mit <code>print()</code> aus.

> <div style="background-color:rgba(0, 0, 250, 0.1)">  
>
> **Erwartete Ausgabe:**
> ````text
>  Gib die Punkte der ersten Prüfung ein: 85
>  Gib die Punkte der zweiten Prüfung ein: 90

>

`````python
# 👉 ---> Dein Code
`````


### ➡️ *Aufgabe:* Mathe 1

Schreibe ein Programm, welches den Benutzer 2 Zahlen (Kommazahlen) eingeben lässt und  
- die Summe,
- die Differenz,
- das Produkt,
- den Quotient und 
- den Divisionsrest  (``Modulo:  zahl1 % zahl2``)

berechnet und ausgibt. Verwende für jede Zahl eine geeignete Variable! (``float``)!

**Wichtiger Hinweis:** Kommazahlen werden im Python-Code und bei der Eingabe immer mit einem **Punkt `.`** geschrieben (z. B. `945.45`), nicht mit einem Komma!

> <div style="background-color:rgba(0, 0, 250, 0.1)">  
>
> **Erwartete Ausgabe:**
> ````text
>   Geben Sie bitte die erste Zahl ein: 945.45
>   Geben Sie bitte die zweite Zahl ein: 1234.567
>
>   945.45 + 1234.567 = 2180.017
>   945.45 - 1234.567 = -289.117
>   945.45 * 1234.567 = 1167221.37015
>   945.45 / 1234.567 = 0.7658153406326123
>   945.45 mod 1234.567 = 945.45
> ````


`````python
# 👉 ---> Dein Code

`````

<details><summary>💡 Lösung</summary>

```python
    # Einlesen und Initialisieren der Variablen
    zahl_1: float = float(input("Geben Sie bitte die erste Zahl ein: "))
    zahl_2: float = float(input("Geben Sie bitte die zweite Zahl ein: "))

    # Berechnen der Ergebnisse
    summe: float = zahl_1 + zahl_2
    differenz: float = zahl_1 - zahl_2
    produkt: float = zahl_1 * zahl_2
    quotient: float = zahl_1 / zahl_2
    rest: float = zahl_1 % zahl_2

    # -------------------------------------------------------------
    # Ausgabe der Variablen
    print()
    print(f"{zahl_1} + {zahl_2} = {summe}")
    print(f"{zahl_1} - {zahl_2} = {differenz}")
    print(f"{zahl_1} * {zahl_2} = {produkt}")
    print(f"{zahl_1} / {zahl_2} = {quotient}")
    print(f"{zahl_1} mod {zahl_2} = {rest}")
```

### ➡️ *Aufgabe:* Der Kreis

Lies den Radius eines Kreises ein.   

Berechne den Durchmesser, den Umfang und die Fläche des Kreises, wobei Pi den Wert 3.14159 hat, und gib das Ergebnis mit <code>print()</code> aus.

- $ durchmesser = 2 \cdot r  $ <br>
- $ umfang  = 2 \cdot r \cdot 3.1415 $ <br>
- $ flaeche = r \cdot r \cdot 3.1415 $ <br>

> <div style="background-color:rgba(0, 0, 250, 0.1)">  
>
> **Erwartete Ausgabe:**
> ````text
>  Gib den Radius des Kreises ein: 7.5
>  
>  Der Radius des Kreises ist: 7.5
>  Der Durchmesser des Kreises ist: 15
>  Der Umfang des Kreises ist: 47.12385
>  Die Fläche des Kreises ist: 176.7144375
> ````

`````python
# 👉 ---> Dein Code
`````

<details><summary>💡 Lösung</summary>

```python
    # Einlesen und Initialisieren der Variablen
    radius: float = float(input("Gib den Radius des Kreises ein: "))
    pi: float = 3.14159

    # Berechnen des Ergebnisses
    durchmesser: float = 2 * radius
    umfang: float = 2 * radius * pi
    flaeche: float = radius * radius * pi

    # -------------------------------------------------------------
    # Ausgabe der Variablen
    print(f"Der Radius des Kreises ist: {radius}")
    print(f"Der Durchmesser des Kreises ist: {durchmesser}")
    print(f"Der Umfang des Kreises ist: {umfang}")
    print(f"Die Fläche des Kreises ist: {flaeche}")
```
