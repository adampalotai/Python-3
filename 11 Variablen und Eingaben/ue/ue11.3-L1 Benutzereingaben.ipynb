### ➡️ *Aufgabe:* Die olympischen Spiele

Lies dein Geburtsjahr und das Jahr der Olympischen Spiele ein.   

Berechne, wie viele Jahre seit dem Jahr deiner Geburt vergangen sind und gib das Ergebnis mit <code>print()</code> aus.  
Gib weiters aus, wie viele Tage diese Jahre sind (ein Jahr sind immer 365 Tage)

> <div style="background-color:rgba(0, 0, 250, 0.1)">  
>
> **Erwartete Ausgabe:**  
> *Verwende jedoch dein tatsächliches Geburtsjahr in diesem Beispiel*
> ````text
>  Gib dein Geburtsjahr ein: 1990
>  Gib das Jahr der Olympischen Spiele ein: 2024
>
>  Geboren bin ich im Jahr 1990
>  Die olympischen Spiele sind im Jahr 2024
>  Es sind seit meiner Geburt 34 Jahre vergangen.
>  Das sind 12410 Tage.
> ````  

`````python
# 👉 ---> Dein Code

bd:int = int(input("Gib dein Geburtsjahr ein: "))
os:int = int(input("Gib das Jahr der Olympischen Spiele ein: 2024"))

print(f"Es sind seit meiner Geburt {os - bd} Jahre vergangen.\nEs sind seit meiner Geburt {(os - bd) * 365} Tage vergangen.")
`````

<details><summary>💡 Lösung</summary>

```python
    # Einlesen und Initialisieren der Variablen
    jahr_geboren: int = int(input("Gib dein Geburtsjahr ein: "))
    jahr_olympiade: int = int(input("Gib das Jahr der Olympischen Spiele ein: "))

    # Berechnen des Ergebnisses
    jahre_vergangen: int = jahr_olympiade - jahr_geboren
    tage_vergangen: int = jahre_vergangen * 365

    # -------------------------------------------------------------
    # Ausgabe der Variablen
    print(f"Geboren bin ich im Jahr {jahr_geboren}")
    print(f"Die olympischen Spiele sind im Jahr {jahr_olympiade}")
    print(f"Es sind seit meiner Geburt {jahre_vergangen} Jahre vergangen.")
    print(f"Das sind {tage_vergangen} Tage.")
```

### ➡️ *Aufgabe:* Die Erde

Schreibe ein Programm, welches berechnet   
- um wie viel der Umfang der Erde sich vergrößert, wenn ihr Durchmesser `um 1m` größer wäre.
  
Der richtige Durchmesser der Erde beträgt `12.742,0 km`  
  
*Achte bei der Berechnung darauf, dass die Änderung in Meter, aber der Durchmesser in Kilometer ist.*
> <div style="background-color:rgba(0, 0, 250, 0.1)">  
>
> **Erwartete Ausgabe:**
> ````text
>   Umfang der Erde bei richtigem Durchmesser: ....  m  
>   Umfang der Erde bei 1m größerem Durchmesser: ....  m

> ````

`````python
# 👉 ---> Dein Code
`````

<details><summary>💡 Lösung</summary>

```python
    # Berechnen des Ergebnisses
    durchmesser: float     = 12742 * 1000                   # Durchmesser auf m umrechnen
    umfang_richtig: float  = durchmesser * 3.14159
    umfang_groesser: float = (durchmesser + 1) * 3.14159
    vergroesserung: float  = umfang_groesser - umfang_richtig

    # -------------------------------------------------------------
    # Ausgabe der Variablen
    print(f"Umfang der Erde bei richtigem Durchmesser: {umfang_richtig} m")
    print(f"Umfang der Erde bei 1m größerem Durchmesser: {umfang_groesser} m")
    print(f"Vergrößerung des Umfangs: {vergroesserung} m")
```

### ➡️ *Aufgabe:* Das Verpackungs-Dilemma

Lies die Anzahl der Äpfel und die Anzahl der Äpfel pro Schachtel ein.  

Berechne, wie viele Schachteln du vollständig füllen kannst und wie viele Äpfel übrig bleiben, da sie nicht für eine weitere volle Schachtel ausreichen.  
Gib das Ergebnis mit <code>print()</code> aus.

> <div style="background-color:rgba(0, 0, 250, 0.1)">  
>
> **Erwartete Ausgabe:**
> ````text
>  Gib die Anzahl der Äpfel ein: 31
>  Gib die Anzahl der Äpfel pro Schachtel ein: 4
>
>  Es sind insgesamt 31 Äpfel vorhanden!
>  In jede Schachtel passen 4 Äpfel!
>  Zum Verpacken werden deshalb 7 Schachteln benöigt!
>  Und es bleiben 3 Äpfel übrig, da die letzte Schachtel nicht mehr voll ist!
> ````

`````python
# 👉 ---> Dein Code
`````

<details><summary>💡 Lösung</summary>

```python
    # Einlesen und Initialisieren der Variablen
    anzahl_aepfel: int = int(input("Gib die Anzahl der Äpfel ein: "))
    anzahl_pro_schachtel: int = int(input("Gib die Anzahl der Äpfel pro Schachtel ein: "))

    # Berechnen des Ergebnisses
    anzahl_schachteln: int = anzahl_aepfel // anzahl_pro_schachtel
    anzahl_rest: int = anzahl_aepfel % anzahl_pro_schachtel

    # -------------------------------------------------------------
    # Ausgabe der Variablen
    print(f"Es sind insgesamt {anzahl_aepfel} Äpfel vorhanden!")
    print(f"In jede Schachtel passen {anzahl_pro_schachtel} Äpfel!")
    print(f"Zum Verpacken werden deshalb {anzahl_schachteln} Schachteln benötigt!")
    print(f"Und es bleiben {anzahl_rest} Äpfel übrig, da die letzte Schachtel nicht mehr voll ist!")
```

### ➡️ *Aufgabe:* Das Eisverkäufer-Dilemma

Du betreibst einen kleinen Eisladen. Lies die Anzahl der Kugeln Eis auf Lager und den Preis pro Kugel ein. 
Lies weiters ein, wie viele Kugeln du in den ersten 2 Stunden verkauft hast und wie viele Kugeln in den nächsten 2 Stunden.

Berechne, wie viel Geld du insgesamt eingenommen hast und wie viele Kugeln Eis noch übrig sind. Gib beides mit <code>print()</code> aus. 

> <div style="background-color:rgba(0, 0, 250, 0.1)">  
>
> **Erwartete Ausgabe:**
> ````text
>  Gib die Anzahl der Kugeln Eis auf Lager ein: 120
>  Gib den Preis pro Kugel ein: 1.5
>  Gib die in den ersten zwei Stunden verkauften Kugeln ein: 35
>  Gib die in den nächsten zwei Stunden verkauften Kugeln ein: 50
>
>  Die Gesamteinnahmen betragen: 127.5 Euro
>  Es sind noch 35 Kugeln Eis übrig.
> ````

`````python
# 👉 ---> Dein Code
`````

<details><summary>💡 Lösung</summary>

```python
    # Einlesen und Initialisieren der Variablen
    kugeln_gesamt: int = int(input("Gib die Anzahl der Kugeln Eis auf Lager ein: "))
    preis_pro_kugel: float = float(input("Gib den Preis pro Kugel ein: "))
    verkauf_erste_stunden: int = int(input("Gib die in den ersten zwei Stunden verkauften Kugeln ein: "))

    # Berechnen des Ergebnisses
    kugeln_verkauft: int = verkauf_erste_stunden + verkauf_zweite_stunden
    einnahmen: float = kugeln_verkauft * preis_pro_kugel
    kugeln_uebrig: int = kugeln_gesamt - kugeln_verkauft

    # -------------------------------------------------------------
    # Ausgabe der Variablen
    print(f"Die Gesamteinnahmen betragen: {einnahmen} Euro")
    print(f"Es sind noch {kugeln_uebrig} Kugeln Eis übrig.")
```
